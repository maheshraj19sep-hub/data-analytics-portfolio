```python
import pandas as pd
import numpy as np

df=pd.read_csv("/content/sample_data/sales_data.csv")
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
df
```

```python
df['sales_date']=pd.to_datetime(df['sales_date'])
df
```

```python
df['month']=df['sales_date'].dt.month
df['year']=df['sales_date'].dt.year
df
```

```python

result=df.groupby('region')['sales_amount'].sum().sort_values(ascending=False)
print(result)
result1=df.groupby('product_category')['sales_amount'].sum().sort_values(ascending=False)
print(result1)


```

```python
import matplotlib.pyplot as plt

chart_data = df.groupby('product_category')['sales_amount'].sum().sort_values(ascending=True)

chart_data.plot(kind='bar', color='orange', edgecolor='black')

plt.title("sales by region and sales rep")
plt.xlabel("region and sales representative")
plt.ylabel("total sales amount ($)")

plt.tight_layout()
plt.show()
```
