```python
import pandas as pd
import numpy as np
```

```python
orders=pd.read_csv("/content/sample_data/olist_orders_dataset.csv")
```

```python
items=pd.read_csv("/content/sample_data/olist_order_items_dataset.csv")
```

```python
products=pd.read_csv("/content/sample_data/olist_products_dataset.csv")
```

```python
customers=pd.read_csv("/content/sample_data/olist_customers_dataset.csv")
```

```python
df=orders.merge(items,on="order_id")
df=df.merge(products,on='product_id')
df=df.merge(customers,on='customer_id')
df
```

```python
df.info()
```

```python
df.head()
```

```python
df['order_purchase_timestamp']=pd.to_datetime(df['order_purchase_timestamp'])
df['order_delivered_customer_date']=pd.to_datetime(df['order_delivered_customer_date'])
df
```

```python
total_revenue=df['price'].sum()
total_orders=df['order_id'].nunique()
aov=total_revenue/total_orders
print("total_revenue =",total_revenue )
print("total_orders",total_orders)
print("aov",aov )
df['delivery_time']=(df['order_delivered_customer_date']-df['order_purchase_timestamp']).dt.days
avg_delivery_time=df['delivery_time'].mean()
```

```python
df['month_year']=df['order_purchase_timestamp'].dt.to_period('m')
monthly_sales=df.groupby('month_year')['price'].sum()
```

```python
df['delivery_time'] = (df['order_delivered_customer_date'] - df['order_purchase_timestamp']).dt.days

avg_delivery_time = df['delivery_time'].mean()
print("avg_delivery_time = ",avg_delivery_time)
```

```python
top_categories=df.groupby('product_category_name')['price'].sum().sort_values(ascending=False).head(10)
print(top_categories)
```

```python
region_sales= df.groupby('customer_state')['price'].sum().sort_values(ascending=False)
print(region_sales)
```

```python
orders_trend=df.groupby('month_year')['order_id'].nunique()
print(orders_trend)
```

```python
late_orders = df[df['delivery_time']>7]
late_percentage = len(late_orders)/len(df)*100
print ("late_orders",late_orders,"late_percentage",late_percentage)
```

```python
monthly_sales.plot(kind='line',title='Monthly Sales')
```

```python
top_categories.plot(kind='bar',title='Top Categories')
```

```python
region_sales.plot(kind='bar',title="Region Sales")
```
